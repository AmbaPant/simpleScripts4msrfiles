```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_globalfit_dat(filename):
    """
    Plot musrfit-style globalfit .dat file:
    xData, data, errData + xTheory, theory
    """
    myleg = ["TF 20 G","ZF","LF 5 G","LF 10 G"] #Here is legends of each data
    mycol = ["black","red","green","blue"] # color of plots
    mymarks = ['o','s','^','d'] # markers of plots
    
    # Read header
    with open(filename) as f:
        header = f.readline().strip()

    cols = [c.strip() for c in header.split(',')]
    data = np.genfromtxt(filename, delimiter=',', skip_header=1)

    #mytext = filename.split("_")[0][:-1] + " K  TF 20 G"
    mytext = filename.split("_")[0] + " K  sampleName" # This for text on plot like T K and sample name
    
    ncols = len(cols)
    nsets = ncols // 5   # each dataset has 5 columns

    #plt.figure(figsize=(8, 6))
    fs = 24
    ms = 8
    lwitdh = 2.5
    
    fig = plt.figure(figsize = [15,10])
    ax = fig.add_subplot(111)

    for axis in ['top', 'bottom', 'left', 'right']:
        ax.spines[axis].set_linewidth(lwitdh)  # change width
        ax.spines[axis].set_color('black')    # change color
    
    plt.xticks(fontsize=fs)
    plt.yticks(fontsize=fs)


    for i in range(nsets):
        xD = data[:, 5*i + 0]
        yD = data[:, 5*i + 1]
        eD = data[:, 5*i + 2]
        xT = data[:, 5*i + 3]
        yT = data[:, 5*i + 4]

        plt.errorbar(
            xD, yD, yerr=eD,
            zorder=0, capsize = 5,color=mycol[i], mfc='white',\
            markeredgewidth=lwitdh,linewidth=lwitdh,linestyle='',marker=mymarks[i], markersize=ms,
            #fmt='o', ms=4, capsize=2,
            label = str(myleg[i]) #you can choose the suitable i for your plots
            #label=f'Data {i}'
        )

        plt.plot(
            xT, yT,
            '-', lw=lwitdh,
            color="blue",
            #color=mycol[i],
            label=''
            #label=f'Theory {i}'
        )
    #plt.axvline(x=4,color='orange')   # vertical line at x = 1.5
    plt.title(filename, fontsize=fs-2)

    plt.xlabel("Time [µs]", fontsize=fs)
    plt.ylabel("Asymmetry", fontsize=fs)
    plt.text(3.0, 0.25, mytext, fontsize=fs)
    plt.axis([-0.1,20,0, 0.3])

    plt.tick_params(axis='both', direction='in', which='major', length=8, width=3)  # Customize major ticks
    plt.tick_params(axis='both', direction='in', which='minor', length=4, width=1.5)   # Customize minor ticks
    #plt.yscale("log")
    plt.minorticks_on()  # Enable minor ticks

    plt.legend(fontsize=fs,loc=1)
    #plt.grid(True)
    plt.tight_layout()
    plt.savefig(filename + ".png")
    plt.show()
```

```python
flist = [
"50_runNo-and-etc.dat",
"50_runNo-and-etc.dat",
"50_runNo-and-etc.dat" #list of your .dat files
]
```

```python
for i in range(len(flist)):
    plot_globalfit_dat(flist[i])
```
