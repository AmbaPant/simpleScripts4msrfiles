```python
import matplotlib.pyplot as plt
import numpy as np
import math
from math import pi
from ambaClassFunctions import *
```

```python
basePath = "C:/Users/pant4/Documents/" #this is path to .dat files

fileList = [
"file1.dat",
"file2.dat",
"file3.dat",
"file4.dat"
]
```

```python
def plotspectra(fileList):
    tempt = fileList.split("_")[0][:-1]

    #msrfile = basePath + fileList[j]
    datfile = basePath + fileList
    pngfile = basePath + fileList+".png"
    figTitle = fileList.split('.')[0]
    print(datfile)

    nrows = 94 # change here for your row numbers
    tobs = readDatfile4msr(datfile,nrows)[0]
    asyobs = readDatfile4msr(datfile,nrows)[1]
    asyerr = readDatfile4msr(datfile,nrows)[2]
    tfit = readDatfile4msr(datfile,nrows)[3]
    asyfit = readDatfile4msr(datfile,nrows)[4]

    gamma_muon = 0.01355387 #MHz/G
    gamma_Mu = 1.3944997 #MHz/G 

    fs = 18 #fontsize
    ms = 8
    
    fig = plt.figure(figsize = [10,8])
    ax = fig.add_subplot(111)

    for axis in ['top', 'bottom', 'left', 'right']:
        ax.spines[axis].set_linewidth(2.0)  # change width
        ax.spines[axis].set_color('black')    # change color

    plt.xticks(fontsize=fs)
    plt.yticks(fontsize=fs)


    plt.errorbar(tobs,asyobs, yerr = asyerr, zorder=1, capsize = 5,color="black", mfc='white',markeredgewidth = 2.0, linestyle='',marker='o', markersize=ms, label = 'Obs')
    plt.plot(tfit, asyfit, color="red", linestyle='-', linewidth=3,label = '')

    plt.text(2.0, 0.18, str(tempt )+' K  TF 20 G', fontsize=fs)

    plt.title(figTitle, fontsize=fs)
    plt.xlabel('Time [$\mu$s]', fontsize=fs)
    plt.ylabel('Asymmetry', fontsize=fs)
    plt.axis([-0.2,15,-0.2, 0.2])
    #plt.grid(True)                                 # comment this if u do not need grid
    plt.tick_params(axis='both', direction='in', which='major', length=8, width=2)  # Customize major ticks
    plt.tick_params(axis='both', direction='in', which='minor', length=4, width=1)   # Customize minor ticks
    #plt.yscale("log")
    plt.minorticks_on()  # Enable minor ticks

    plt.legend(fontsize=fs)

    plt.tight_layout()                             # this is for margin
    plt.savefig(pngfile)
    plt.show()
```

```python
i = 0
while i < (len(fileList)):
    plotspectra(fileList[i])
    i = i + 1
print("All done !! ")
```
