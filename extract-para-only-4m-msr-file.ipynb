```python
def extractChisq(fname):
    #print(fname)
    if exists(fname):
        for line in open(fname, 'r'):
            each_line = line.split(None, 10)
            if len(each_line) < 5:
                continue
            else:
                needed_part = each_line
                #print(needed_part)
                parameter = 'chisq'
                tot_let = len( parameter )
                if needed_part[0][:tot_let] == parameter:
                    ser_no = needed_part[2][:-1]
                    #print(needed_part)
                    chi_val = needed_part[8]
                    return(chi_val)
```

```python
def extractPara(msrfile,outfile):
    #adjust the max_rows and skiprows
    # 
    #a = np.loadtxt(msrfile, skiprows=4, max_rows=10,dtype = str, comments='%') #comments=None
    #a = np.loadtxt(msrfile, skiprows=4, max_rows=5,dtype = str) #comments=None , comments='%' this for TF20 only
    a = np.loadtxt(msrfile, skiprows=4, max_rows=8,dtype = str) #comments=None , comments='%' this for TF20 two funs
    #fileList = np.genfromtxt(datfile, delimiter=',', dtype=str, comments='%', skip_header=4, invalid_raise=False)

    with open(outfile,'a+') as fw:
        T = float(msrfile.split("_")[0][:-1]) #[2]) #[0][:-1])
        #print(T)
        fieldVal = re.findall(r"\d+", msrfile)[1]
        
        fw.write('\n %s  %s\t' %(msrfile,fieldVal))

    with open(outfile, 'a+') as f:
        for i in range(len(a)):
            f.write("%s %s %s\t" %(a[i][1],a[i][2],a[i][3]))

    with open(outfile,'a+') as fc:
        fc.write("%s %s\t" %("chiSq",extractChisq(msrfile)))
```
