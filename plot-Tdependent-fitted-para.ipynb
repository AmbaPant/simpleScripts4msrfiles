```python
def minval(myf,n):
    data = np.loadtxt(myf, dtype = str)
    #print(data)
    minval =  np.min(data[:,n].astype(float))
    return(minval)
```

```python
def avgval(myf,n):
    data = np.loadtxt(myf, dtype = str)
    #print(data)
    avgval =  np.mean(data[2:,n].astype(float))
    return(avgval)
```

```python
#
# delta B = B - B300K
# err (delta B) = sqrt (errB^2 + errB300K)
# H2O B 300K, fieldTF 20.647 0.012
# D2O B 300K, fieldTF 20.887 0.014
```

```python
def plotBdiff(filename,rowi,rowf,xcol,ycol, mymarker, mylstyle, mymarkcolor,myfillcolor, myleg):

    data = np.loadtxt(filename, dtype = str)
    
    x =  data[rowi:rowf,xcol].astype(float)
    print(x)
    B300K = data[0:1,ycol].astype(float)
    errB300K = data[0:1,ycol+1].astype(float)
    print(B300K)
    print(errB300K)
    
    y =  np.abs(data[rowi:rowf,ycol].astype(float) - B300K)
    print(y)
    erry =  ((data[rowi:rowf,ycol+1].astype(float))**2 + (errB300K)**2)**0.5
    print(erry)
    plt.errorbar(x, y, yerr = erry, zorder=0, capsize = 5,color=mymarkcolor, \
                 mfc=myfillcolor, markeredgewidth=3,linewidth=3,linestyle=mylstyle,marker=mymarker, markersize=ms, label = myleg)
```
